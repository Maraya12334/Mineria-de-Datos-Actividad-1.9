```python
df = df.drop_duplicates().copy()

print("Registros después de eliminar duplicados:", len(df))
```

```python
faltantes = df.isnull().sum()
faltantes = faltantes[faltantes > 0].sort_values(ascending=False)

faltantes
```

```python
ratings_validos = ["E", "E10+", "T", "M"]

df_arbol = df[df["Rating"].isin(ratings_validos)].copy()

print("Registros para el árbol:", len(df_arbol))
print("\nDistribución de Rating:")
print(df_arbol["Rating"].value_counts())
```

```python
df_arbol["Metascore"] = pd.to_numeric(
    df_arbol["Metascore"],
    errors="coerce"
)

print("Tipo de dato de Metascore:", df_arbol["Metascore"].dtype)
print("Valores faltantes en Metascore:", df_arbol["Metascore"].isnull().sum())
```

```python
df_arbol["Userscore"] = pd.to_numeric(
    df_arbol["Userscore"],
    errors="coerce"
)

print("Tipo de dato de Userscore:", df_arbol["Userscore"].dtype)
print("Valores faltantes en Userscore:", df_arbol["Userscore"].isnull().sum())
```

```python
df_arbol["Metascore_Categoria"] = pd.cut(
    df_arbol["Metascore"],
    bins=[-1, 59, 79, 100],
    labels=["Bajo", "Medio", "Alto"]
)

print(df_arbol["Metascore_Categoria"].value_counts(dropna=False))
```

```python
df_arbol["Userscore_Categoria"] = pd.cut(
    df_arbol["Userscore"],
    bins=[-0.1, 4.9, 7.4, 10],
    labels=["Bajo", "Medio", "Alto"]
)

print(df_arbol["Userscore_Categoria"].value_counts(dropna=False))
```

```python
df_arbol["CriticReviews_Categoria"] = pd.cut(
    df_arbol["# of Critic Reviews"],
    bins=[-1, 9, 29, np.inf],
    labels=["Pocas", "Medias", "Muchas"]
)

print(df_arbol["CriticReviews_Categoria"].value_counts(dropna=False))
```

```python
genres = df_arbol["Genres"].fillna("").astype(str).str.split(", ")

print("Ejemplo de géneros separados:")
print(genres.head())
```

```python
genres_dummies = genres.str.join("|").str.get_dummies(sep="|")

print("Cantidad de categorías de género:", genres_dummies.shape[1])
print("\nPrimeras filas:")
genres_dummies.head()
```

```python
rating_dummies = pd.get_dummies(
    df_arbol["Rating"],
    prefix="Rating"
)

metascore_dummies = pd.get_dummies(
    df_arbol["Metascore_Categoria"],
    prefix="Metascore"
)

userscore_dummies = pd.get_dummies(
    df_arbol["Userscore_Categoria"],
    prefix="Userscore"
)

critic_dummies = pd.get_dummies(
    df_arbol["CriticReviews_Categoria"],
    prefix="CriticReviews"
)

print("Variables binarias creadas correctamente.")
print("Rating:", rating_dummies.shape[1])
print("Metascore:", metascore_dummies.shape[1])
print("Userscore:", userscore_dummies.shape[1])
print("Critic Reviews:", critic_dummies.shape[1])
```

```python
datos_items = pd.concat(
    [
        genres_dummies,
        rating_dummies,
        metascore_dummies,
        userscore_dummies,
        critic_dummies
    ],
    axis=1
)

datos_items = datos_items.astype(bool)

print("Filas:", datos_items.shape[0])
print("Columnas:", datos_items.shape[1])
```

```python
from mlxtend.frequent_patterns import apriori, association_rules

print("Herramientas para reglas de asociación cargadas correctamente.")
```

```python
frecuentes_10 = apriori(
    datos_items,
    min_support=0.10,
    use_colnames=True
)

print("Cantidad de conjuntos frecuentes:", len(frecuentes_10))
```

```python
frecuentes_10.sort_values(
    "support",
    ascending=False
).head(15)
```

```python
for soporte in [0.30, 0.20, 0.10]:
    conjuntos = apriori(
        datos_items,
        min_support=soporte,
        use_colnames=True
    )

    print(
        f"min_support = {soporte}: "
        f"{len(conjuntos)} conjuntos frecuentes"
    )
```

```python
reglas_10 = association_rules(
    frecuentes_10,
    metric="confidence",
    min_threshold=0.30
)

print("Cantidad de reglas generadas:", len(reglas_10))
```

```python
for confianza in [0.30, 0.50, 0.70]:
    reglas = association_rules(
        frecuentes_10,
        metric="confidence",
        min_threshold=confianza
    )

    print(
        f"min_confidence = {confianza}: "
        f"{len(reglas)} reglas"
    )
```

```python
reglas_70 = association_rules(
    frecuentes_10,
    metric="confidence",
    min_threshold=0.70
)

reglas_70[
    ["antecedents", "consequents", "support", "confidence", "lift"]
].sort_values(
    "confidence",
    ascending=False
).head(15)
```

```python
for lift_min in [1.0, 1.2, 1.5]:
    reglas_lift = reglas_70[
        reglas_70["lift"] >= lift_min
    ]

    print(
        f"min_lift = {lift_min}: "
        f"{len(reglas_lift)} reglas"
    )
```

```python
reglas_finales = reglas_70[
    ["antecedents", "consequents", "support", "confidence", "lift"]
].sort_values(
    "lift",
    ascending=False
)

reglas_finales.head(10)
```

```python
reglas_seleccionadas = reglas_70[
    (
        (reglas_70["antecedents"].apply(lambda x: "First-Person" in x and "Action" in x))
        & (reglas_70["consequents"].apply(lambda x: "Shooter" in x))
    )
    |
    (
        (reglas_70["antecedents"].apply(lambda x: "Arcade" in x and "Action" in x))
        & (reglas_70["consequents"].apply(lambda x: "Shooter" in x))
    )
    |
    (
        (reglas_70["antecedents"].apply(lambda x: "Real-Time" in x))
        & (reglas_70["consequents"].apply(lambda x: "Strategy" in x))
    )
][["antecedents", "consequents", "support", "confidence", "lift"]]

reglas_seleccionadas
```

```python
tabla_reglas = reglas_seleccionadas.copy()

tabla_reglas["support_%"] = (tabla_reglas["support"] * 100).round(2)
tabla_reglas["confidence_%"] = (tabla_reglas["confidence"] * 100).round(2)
tabla_reglas["lift"] = tabla_reglas["lift"].round(2)

tabla_reglas[
    ["antecedents", "consequents", "support_%", "confidence_%", "lift"]
]
```

### Interpretación de las reglas de asociación

Se seleccionaron tres reglas utilizando `min_support = 0,10`, `min_confidence = 0,70` y un `lift` mínimo de 1,5. Estos parámetros permitieron reducir la cantidad de reglas y concentrarse en relaciones con una confianza elevada.

**1. Real-Time → Strategy**

Esta regla presenta un soporte de **11,01%**, una confianza de **100%** y un lift de **4,99**. Esto significa que el 11,01% de los videojuegos analizados contiene la característica `Real-Time` y, dentro de esos casos, todos también presentan la característica `Strategy`. El lift de 4,99 indica que la presencia conjunta de ambas características es considerablemente mayor que la esperada si fueran independientes.

**2. Arcade + Action → Shooter**

La regla presenta un soporte de **12,12%**, una confianza de **98,11%** y un lift de **5,82**. Esto indica que, cuando un videojuego presenta simultáneamente las características `Arcade` y `Action`, en el 98,11% de los casos también aparece `Shooter`. El lift muestra una asociación positiva importante entre estas características.

**3. First-Person + Action → Shooter**

Esta regla presenta un soporte de **12,50%**, una confianza de **99,07%** y un lift de **5,88**. Esto significa que, cuando un videojuego presenta las características `First-Person` y `Action`, en el 99,07% de los casos también presenta `Shooter`. El lift de 5,88 indica una asociación positiva fuerte entre las características.

### Utilidad para la toma de decisiones

Las reglas permiten identificar combinaciones de géneros y características que aparecen frecuentemente en los videojuegos del dataset. Esta información podría utilizarse para apoyar la clasificación de nuevos videojuegos, organizar catálogos por características relacionadas o detectar combinaciones habituales de géneros. Las reglas muestran asociaciones en los datos, pero no implican necesariamente una relación causal entre las características.


```python
print("Transformación de variables numéricas:")
print("- Metascore → categorías Bajo, Medio y Alto")
print("- Userscore → categorías Bajo, Medio y Alto")
print("- # of Critic Reviews → categorías Pocas, Medias y Muchas")
```

```python
variables_numericas = [
    "Metascore",
    "Userscore",
    "# of Critic Reviews",
    "Critic Positive",
    "Critic Mixed",
    "Critic Negative",
    "# of User Reviews",
    "User Positive",
    "User Mixed",
    "User Negative"
]

df_arbol[variables_numericas].isnull().sum()
```

```python
X = df_arbol[variables_numericas].copy()
y = df_arbol["Rating"].copy()

# Completar valores faltantes con la mediana de cada variable
X = X.fillna(X.median())

print("Datos preparados para el árbol.")
print("Filas:", X.shape[0])
print("Variables predictoras:", X.shape[1])
print("Valores faltantes restantes:", X.isnull().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Datos de entrenamiento:", X_train.shape[0])
print("Datos de prueba:", X_test.shape[0])
print("\nDistribución del objetivo en entrenamiento:")
print(y_train.value_counts())
print("\nDistribución del objetivo en prueba:")
print(y_test.value_counts())
```

```python
from sklearn.tree import DecisionTreeClassifier

arbol = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

arbol.fit(X_train, y_train)

print("Árbol de decisión entrenado correctamente.")
print("Profundidad máxima:", arbol.get_depth())
print("Cantidad de hojas:", arbol.get_n_leaves())
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = arbol.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy del árbol:", round(accuracy, 4))
print("\nReporte de clasificación:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(22, 12))

plot_tree(
    arbol,
    feature_names=X.columns,
    class_names=arbol.classes_,
    filled=True,
    rounded=True,
    fontsize=9
)

plt.title("Árbol de decisión para clasificar Rating")
plt.show()
```

```python
print("Variable raíz:", X.columns[arbol.tree_.feature[0]])
print("Punto de corte:", round(arbol.tree_.threshold[0], 2))
```

```python
from sklearn.tree import export_text

reglas_arbol = export_text(
    arbol,
    feature_names=list(X.columns)
)

print(reglas_arbol)
```

```python
importancias = pd.Series(
    arbol.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

print(importancias)
```

### Interpretación del árbol de decisión

Se construyó un árbol de decisión para clasificar los videojuegos según su clasificación `Rating`: **E, E10+, T y M**. Para evitar una estructura excesivamente compleja, se estableció una profundidad máxima de **4 niveles**.

El conjunto de datos se dividió en **2738 registros para entrenamiento** y **685 para prueba**. El modelo obtuvo un **accuracy de 43,07%** sobre los datos de prueba. El resultado muestra que el árbol presenta dificultades para distinguir algunas categorías, especialmente `E10+`, que no fue predicha correctamente en el conjunto de prueba. Este resultado debe considerarse al interpretar las reglas obtenidas.

### Variable raíz

La variable utilizada como raíz fue **`# of User Reviews`**, con un punto de corte de **118,5 reseñas**.

Esto significa que la primera división realizada por el árbol separa los videojuegos en dos grupos:

* `# of User Reviews ≤ 118,5`
* `# of User Reviews > 118,5`

Además, esta variable presentó la mayor importancia dentro del árbol, con un valor de **0,468450**, por lo que fue el predictor que más contribuyó a las decisiones del modelo.

### Camino 1 — Regla de negocio

Un camino del árbol permite establecer la siguiente regla:

**Si `# of User Reviews ≤ 118,5`, `Critic Mixed ≤ 8,5`, `# of User Reviews ≤ 38,5` y `Metascore ≤ 57,5`, entonces el modelo clasifica el videojuego como `T`.**

Esta regla puede utilizarse como una forma de clasificación basada en la cantidad de reseñas de usuarios, la cantidad de críticas mixtas y el Metascore.

### Camino 2 — Regla de negocio

Otro camino permite establecer:

**Si `# of User Reviews > 118,5`, `User Mixed ≤ 30,5`, `Metascore > 81,5` y `Userscore ≤ 7,55`, entonces el modelo clasifica el videojuego como `E`.**

Esta regla combina la cantidad de reseñas de usuarios con indicadores de valoración de críticos y usuarios para determinar la clasificación estimada.

### ¿Por qué utilizar una profundidad máxima de 4?

Se utilizó una profundidad máxima de **4** porque la pauta de la actividad establece este límite y porque permite obtener un árbol relativamente fácil de interpretar. Un árbol mucho más profundo podría generar más reglas y divisiones específicas, aumentando la complejidad y el riesgo de sobreajuste.

### Decisión apoyada por el árbol

El árbol permite apoyar una clasificación inicial de videojuegos según su `Rating`, utilizando características relacionadas con las reseñas y puntuaciones de usuarios y críticos. Sin embargo, debido al accuracy de **43,07%**, las predicciones deben interpretarse como un apoyo exploratorio y no como un sistema definitivo de clasificación.
